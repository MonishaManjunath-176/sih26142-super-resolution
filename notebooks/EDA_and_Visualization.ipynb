### 1. Initialize U-Net Generator Model Architecture (4x Upscaling)

```python
# Create Generator: Input 10m (4 channels RGB+NIR) -> Output 2.5m (4 channels)
generator = UNetGenerator(in_channels=4, out_channels=4, base_features=64, upscale_factor=4)
test_input = torch.rand(1, 4, 64, 64)  # 64x64 at 10m spatial resolution
test_output = generator(test_input)     # 256x256 at 2.5m spatial resolution

print(f"Input shape (10m LR): {test_input.shape}")
print(f"Output shape (2.5m SR): {test_output.shape}")
assert test_output.shape == (1, 4, 256, 256), "Output shape check failed!"
```
